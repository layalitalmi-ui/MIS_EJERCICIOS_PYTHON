```python

rey_blanco: str = "\u2654"              # --------------------------------------------------
rey_negro : str = "\u265A"              
peon_negro : str = "\u265F"             # --------------------------------------------------
peon_blanco :str = "\u2659"             
alfil_blanco : str = "\u2657"           # --------------------------------------------------  
alfil_negro : str = "\u265D"            # DEFINICIÓN DE LAS PIEZAS DE AJEDREZ
dama_negra : str = "\u265B"             # Se asigna a cada pieza su símbolo Unicode correspondiente
dama_blanca : str = "\u2655"            # El tipo str indica que cada pieza se representa mediante una cadena de texto.
c_negro : str = "\u265E"
c_blanco : str = "\u2658"               # --------------------------------------------------
t_blanca : str = "\u2656"
t_negra : str = "\u265C"                # --------------------------------------------------

tablero : list =[]                      # ESTABLECER LISTA TABLERO
                                        #Creamos una lista vacía llamada tablero , que próximamente le añadiremos columnas y filas

for i in range (9) :                    # esta línea hace que se creen 9 filas 
    fila_vacia : list = []              # creamos una lista vacía que representará una fila
    for j in range (9):                 # dentro del primer for, otro for que crea 9 casillas en cada fila
        fila_vacia.append("")           # añadimos a las filas vacías espacios qeu simulan las casillas del tablero
    tablero.append(fila_vacia)          # añadimos fila_vacia a la lista vacía , tablero

tablero[0] = [".", "a ", "b ", "c ", "d ", "e ", "f ", "g ", "h "]  # Establecemos que la primera fila del tablero 9x9 tendrá este aspecto

for i in range (9):                          # recorremos las 9 filas del tablero
    for j in range (9):                      # recorremos las 9 columnas del tablero
        print(tablero[i][j], end="    ")     # mostramos el tablero con sus filas y , columnas separadas a una cierta distancia " end="  " "
    print()                                  # cuando termine la primera fila se establece dos espacios entre cada una
    print()
    tablero[8][5] = rey_blanco               # --------------------------------------------------
    tablero[8][4] = dama_blanca
    tablero[8][3]= alfil_blanco              # --------------------------------------------------
    tablero[8][2]= c_blanco                  # PIEZAS DEL TABLERO
    tablero[8][1]= t_blanca                  # ponemos las piezas blancas de ajedrez en la posición de inicio , estas no se pueden establecer mediante un for por que cada posicion es diferente
    tablero[8][6]= alfil_blanco              # --------------------------------------------------
    tablero[8][7]= c_blanco                   
    tablero[8][8]= t_blanca                  # --------------------------------------------------

    for columna in range (1,9):              # creamos un for que vaya de la posicion 1 a 8
        tablero[7][columna]=peon_blanco      # recorremos la fila siete in range de 1 a 8 del tablero llenando los espacios de peones blancos

    tablero[1][5] = rey_negro                # --------------------------------------------------
    tablero [1][4]= dama_negra               
    tablero[1][3]= alfil_negro               # --------------------------------------------------
    tablero[1][2]= c_negro                   # PIEZAS DEL TABLERO
    tablero[1][1]= t_negra                   # ponemos las piezas negras de ajedrez en la posición de inicio , estas no se pueden establecer mediante un for por que cada posicion es diferente
    tablero[1][6]= alfil_negro               # --------------------------------------------------
    tablero[1][7]= c_negro                   
    tablero[1][8]= t_negra                   # --------------------------------------------------

    for columna in range (1,9):              # creamos un for que vaya de la posicion 1 a 8
        tablero[2][columna]=peon_negro       # recorremos la fila dos in range de 1 a 8 del tablero llenando los espacios de peones negros

    for var in range (9):                    # creamos un for que recorra del 0 al 8
            tablero[var][0]= 9 - var         # recorremos la primera columna del tablero de la posición 1 a la 8 , y cambiamos los espacion por números del 8 al 1, restando 9 menos el for que recorrimos del 1 al 8.



columnas_dict = {".":0, "a ": 1, "b ": 2, "c ": 3, "d ": 4, "e ": 5, "f ": 6, "g ": 7, "h ": 8,}  # creamos un diccionario para establecer el valor de cada letra de la primera fila del tablero , para que cuando el jugador introduzca su casilla , sea una letra y un número , y python lo pueda leer y ejecutar.

filas_dict = {8: 1, 7: 2, 6: 3, 5: 4, 4: 5, 3: 6, 2: 7, 1: 8} # otro diccionario para los numeros de la primera columna , ya que están invertidos con el orden que les dimos

nombres_piezas = {"" : "casilla vacía" , "♟" : "peón negro", "♜": "torre negra", "♞": "caballo negro", "♝": "alfil negro", "♛": "dama negra",
     "♚": "rey negro",
     "♙": "peón blanco",
    "♖": "torre blanca",
    "♘": "caballo blanco",
    "♗": "alfil blanco",
    "♕": "dama blanca",
    "♔": "rey blanco" }          # aqui creamos otro diccionario para que cuando tengamos que dar la respuesta , podamos decir el nombre y no solo printear que hay en esa cierta casilla


estado: str = "turno_blancas" # SIEMPRE empieza asi , se establece que empiezan las piezas blancas como en una partida de asedrez , a la variable estado


while estado != "salir":      # Introducimos un while para que la acción se repita hasta que estado sea igual a "salir", es decir que se cierre el bucl
    
    if estado == "turno_blancas":     # SI EL ESTADO = TURNO DE BLANCAS
        print("Turno de blancas")     # mostramos un mensaje que indique que empiezan las blancas

        casilla = input("Casilla a consultar (o 'salir' para terminar):") # preguntamos la casilla mediante un input
        if casilla == "salir":                   # --------------------------------------------------
            estado: str = "salir"                # si se introduce "salir" el bucle se para
            print("Casilla no válida")
            break                                # --------------------------------------------------
        if len(casilla) == 0 or len(casilla) > 2 :  # si la longitud de la casilla es mayor que dos o igual a 0 se printea un mensaje y se continua el bucle
            print("Casilla no válida")
            continue
        
                
        l = casilla[0] + " "       # seleccionamos la primera primera parte de la casilla introducida mas un espacio ya que en el diccionaria de las letras , se han escrito con espacios
        n = casilla[1:]            # seleccionamos la segunda parte de casilla (que es un número)

        if l not in columnas_dict :      # establecemos que si la letra seleccionada de casilla no esta en el diccionario de las columnas que mande un mensaje y continue
            print("Casilla no válida")
            continue

        num = int(n)         # establecemos en una variable num , que el numero seleccionado de casilla tiene que ser un numero entero

        if num not in filas_dict :       # si num no esta en el diccionario de las filas , que mand eun mensaje y countinue el bucle
            print("Casilla no válida")
            continue


 
        if num in filas_dict and l in columnas_dict :    #  si lo anterior no se cumple , es decir , que num esté en el diccionario de las filas y l esté en el diccionario de las columnas , pasamos a mostrar lo que el jugador ha introducido
            fila = filas_dict[num]                       # llamamos fila a el numero seleccionado del diccionario de filas
            columna = columnas_dict[l]                   # llamamos columna a la la letra seleccionada del diccionario de las columnas
            pieza = tablero[fila][columna]               # establecemos una variable llamada pieza que es igual a la casilla seleccionada del tablero mediante las variables fila y columna establecidas anteriormente
            pieza_print = nombres_piezas[pieza]          # establecemos pieza_print como el nombre escogido por pieza dentro del diccionario del nombre de las piezas , nombres_piezas
            print("Hay un" , pieza_print ,"en la casilla", casilla)  # mostramos el mensaje con la variable pieza_print y con la casilla introducida por el jugador
            estado = "turno_negras"  # cambiamos estado a "turno_negras"



        else :                           # si lo anterior no es cierto establecemos el estado a "turno_negras"
            estado = "turno_negras"

                                        # EL MISMO PROCEDIMIENTO QUE EN EL TURNO DE LAS BLANCAS
    elif estado == "turno_negras":
        print("Turno de negras")
        
        casilla = input("Casilla a consultar (o 'salir' para terminar):")
        if casilla == "salir":   
            estado: str = "salir"
            print("Casilla no válida")
            break
        if len(casilla) == 0 or len(casilla) > 2 :
            print("Casilla no válida")
            continue
        
        
        l = casilla[0] + " "
        n = casilla[1:]

        if l not in columnas_dict :
            print("Casilla no válida")
            continue
        num = int(n)

        if num not in filas_dict :
            print("Casilla no válida")
            continue

        if num in filas_dict and l in columnas_dict :
            fila = filas_dict[num]
            columna = columnas_dict[l]
            pieza = tablero[fila][columna]
            pieza_print= nombres_piezas[pieza]
            print("Hay un" , pieza_print ,"en la casilla", casilla)
            estado = "turno_blancas"


        else :
            estado = "turno_blancas"
```
